# Unidad de Medida: `RENDIMIENTO_T_HA`

### 1. Definición
* **Variable:** `RENDIMIENTO_T_HA` (tipo numérico continuo, `float64`).
* **Unidad base:** **Toneladas métricas por hectárea ($t/ha$)**.
* **Significado físico:** Tasa de productividad intensiva que mide la masa de grano de cebada obtenida por cada $10,000\text{ m}^2$ de cultivo, independiente del tamaño de la parcela.

---

### 2. Factores de Conversión
* **A kilogramos por hectárea:**
  $$\text{kg/ha} = \text{RENDIMIENTO\_T\_HA} \times 1{,}000$$
* **A gramos por metro cuadrado (biomasa directa de grano):**
  $$\text{g/m}^2 = \text{RENDIMIENTO\_T\_HA} \times 100$$
* **Producción bruta total del predio:**
  $$\text{Producción Total (t)} = \text{RENDIMIENTO\_T\_HA} \times \text{AREA\_HA}$$

---

# Vamos a tomar en cuenta el data set completo ahora

```python
FECHA_MIN = '2025-04-01'
FECHA_MAX = '2025-10-31'
UMBRAL_NUBES = 30.0

df1_qc = df1[
    (df1['fecha_captura'] >= FECHA_MIN) & 
    (df1['fecha_captura'] <= FECHA_MAX) & 
    (df1['porcentaje_nubosidad'].fillna(100) <= UMBRAL_NUBES)
].copy()

df2_qc = df2[
    (df2['fecha_captura'] >= FECHA_MIN) & 
    (df2['fecha_captura'] <= FECHA_MAX) & 
    (df2['porcentaje_nubosidad'].fillna(100) <= UMBRAL_NUBES)
].copy()
```

```python
s2_qc = df1_qc[(df1_qc['sensor'] == 'Sentinel-2') & (df1_qc['ndvi_promedio'] > 0.05)].copy()
landsat_qc = df1_qc[(df1_qc['sensor'] == 'Landsat') & (df1_qc['vi6t_promedio'].notna())].copy()
planet_qc = df2_qc[df2_qc['ndvi_promedio'] > 0.05].copy()
```

```python
def agregar_metricas(df, prefijo, variables):
    aggs = {}
    for var in variables:
        aggs[f'{prefijo}_{var}_max'] = (var, 'max')
        aggs[f'{prefijo}_{var}_mean'] = (var, 'mean')
        aggs[f'{prefijo}_{var}_integral'] = (var, 'sum')
        aggs[f'{prefijo}_{var}_std'] = (var, 'std')
    return df.groupby('ID_POLIGONO').agg(**aggs).reset_index()
```

```python
feat_s2 = agregar_metricas(s2_qc, 's2', ['ndvi_promedio', 'evi_promedio', 'savi_promedio', 'ndwi_promedio'])
feat_planet = agregar_metricas(planet_qc, 'planet', ['ndvi_promedio', 'evi_promedio', 'lai_promedio', 'msavi_promedio'])
feat_landsat = agregar_metricas(landsat_qc, 'landsat', ['vi6t_promedio'])
```

```python
df_sensores = y.merge(feat_planet, on='ID_POLIGONO', how='left')
df_sensores = df_sensores.merge(feat_s2, on='ID_POLIGONO', how='left')
df_sensores = df_sensores.merge(feat_landsat, on='ID_POLIGONO', how='left')
```
